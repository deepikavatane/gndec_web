```python
class PERSON:
    def __init__(self, name, age, blood, nationality):
        self.name = name
        self.age = age
        self.blood = blood
        self.nationality = nationality


class STUDENT(PERSON):
    def __init__(self, name, age, blood, nationality,
                 school_college, percentage, section, roll_number):
        super().__init__(name, age, blood, nationality)
        self.school_college = school_college
        self.percentage = percentage
        self.section = section
        self.roll_number = roll_number

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Blood Group:", self.blood)
        print("Nationality:", self.nationality)
        print("School/College:", self.school_college)
        print("Percentage:", self.percentage)
        print("Section:", self.section)
        print("Roll Number:", self.roll_number)


student1 = STUDENT("Rahul", 20, "B+", "Indian",
                   "ABC College", 85.5, "A", 101)

student1.display()
```

```python
class Student:
    # Public
    name = ""
    dept = ""
    clg = ""
    year = 0

    # Private
    __regno = 0
    __percentage = 0
    __section = ""
    __father = ""
    __address = ""

    def setdata(self):
        self.__regno = 101
        self.__percentage = 85
        self.__section = "A"
        self.__father = "Ramesh"
        self.__address = "Bangalore"

    def show(self):
        print(self.name, self.dept, self.clg, self.year)
        print(self.__regno, self.__percentage)
        print(self.__section, self.__father, self.__address)


s = Student()
s.name = "Ravi"
s.dept = "CSE"
s.clg = "ABC College"
s.year = 2

s.setdata()
s.show()
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# Concrete class
class CreditCardPayment(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


class UPIPayment(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


# Using the project
payment1 = CreditCardPayment()
payment1.pay(1000)

payment2 = UPIPayment()
payment2.pay(500)
```
